```python
import pandas as pd
from pandas.tseries.offsets import DateOffset
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import griddata
import datetime
from pathlib import Path
import cartopy.crs as ccrs                 # For plotting maps
import cartopy.feature as cfeature         # For plotting maps
from cartopy.util import add_cyclic_point  # For plotting maps
```

```python
#Set working Directory for atm chem
result_dir_atm = Path('/glade/derecho/scratch/demurray/archive/FCnudged_f09.mam.Murray.Mar18_02.2000_2021.001/atm/hist')

#Read in monthly files from current run (fincl1)
files_to_open = ['FCnudged_f09.mam.Murray.Mar18_02.2000_2021.001.cam.h0.2002-01.nc', 
                 'FCnudged_f09.mam.Murray.Mar18_02.2000_2021.001.cam.h0.2002-06.nc']
file_paths = [result_dir_atm / filename for filename in files_to_open]
nc_load = xr.open_mfdataset(file_paths, combine='nested')
```

```python
nc_dep = nc_load[['WD_NH4', 'WD_HNO3', 'WD_NH3']]
nc_dep['sign_WD_NH4'] = xr.where(nc_dep['WD_NH4'] > 0, 1, -1)
nc_dep['sign_WD_NH3'] = xr.where(nc_dep['WD_NH3'] > 0, 1, -1)
nc_dep['WD_NH4'] = nc_dep['WD_NH4']*(86400*30)*1000000
nc_dep['WD_HNO3'] = nc_dep['WD_HNO3']*(86400*30)*1000000
#nc_dep
```

```python
nc_dep['WD_NH4'].max().values
```

```python
#make a map setup:
var_sel = nc_dep['WD_NH4']
print(var_sel)

#select the variable at a specific time 
var_srf = var_sel.isel(time=0)
print(var_srf.shape)

#extract grid variables
lat = var_sel.coords['lat']
lon = var_sel.coords['lon']
print(lat.shape)

var_srf_cyc, lon_cyc = add_cyclic_point(var_srf, coord=lon)

plt.figure(figsize=(20,8))

#Define projection
ax = plt.axes(projection=ccrs.PlateCarree())

#define contour levels
#data_min = var_sel.min().values
#data_max = var_sel.max().values
#num_levels = 10
#clev = np.linspace(data_min, data_max, num_levels)
clev = np.arange(-100,100,1)

#plot the data
plt.contourf(lon_cyc,lat,var_srf_cyc,clev,cmap='Spectral_r',extend='both')

# add coastlines
ax.add_feature(cfeature.COASTLINE)

#add lat lon grids
gl = ax.gridlines(draw_labels=True, color='grey', alpha=0.8, linestyle='--')
gl.xlabels_top = False
gl.ylabels_right = False

# Titles
# Main
plt.title("Global map of Cam-Chem WD_NH4 mg/m2/month, January 2002",fontsize=18)

# y-axis
ax.text(-0.04, 0.5, 'Latitude', va='bottom', ha='center',
        rotation='vertical', rotation_mode='anchor',
        transform=ax.transAxes)
# x-axis
ax.text(0.5, -0.08, 'Longitude', va='bottom', ha='center',
        rotation='horizontal', rotation_mode='anchor',
        transform=ax.transAxes)
# legend
ax.text(1.15, 0.5, 'WD_NH4 mg/m2/month', va='bottom', ha='center',
        rotation='vertical', rotation_mode='anchor',
        transform=ax.transAxes)

plt.colorbar()
plt.show()
#plt.savefig('/path/to/your/folder/figure.png')
```

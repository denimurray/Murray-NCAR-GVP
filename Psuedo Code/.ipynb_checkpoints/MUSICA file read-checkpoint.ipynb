```python
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from pathlib import Path
import os
import pandas as pd
from tqdm import tqdm
from scipy.spatial.distance import cdist
```

```python
# Function to calculate Haversine distance
def haversine(lat1, lon1, lat2, lon2):
    R = 6371.0  # approximate radius of Earth in km
    lat1_rad = np.radians(lat1)
    lon1_rad = np.radians(lon1)
    lat2_rad = np.radians(lat2)
    lon2_rad = np.radians(lon2)

    dlon = lon2_rad - lon1_rad
    dlat = lat2_rad - lat1_rad

    a = np.sin(dlat / 2)**2 + np.cos(lat1_rad) * np.cos(lat2_rad) * np.sin(dlon / 2)**2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))

    distance = R * c
    return distance
```

```python
#Read in the MUSCIA trial files
result_dir =  Path("/glade/derecho/scratch/demurray/archive/MUSICA_trial_050824_01/atm/hist/")
file = "MUSICA_trial_050824_01.cam.h0.2017-01-01-00000.nc"
file_to_open = result_dir / file
nc_load = xr.open_dataset(file_to_open, engine='netcdf4')
nc_load = nc_load[['lat', 'lon']] 
```

```python
#Subset lat and lon for only over north america
clipped_lat = nc_load['lat'].clip(min=24, max=50)
clipped_lon = nc_load['lon'].clip(min=235, max=294)

# Step 2: Update the dataset with clipped variables
nc_load_clipped = nc_load.assign(lat=clipped_lat, lon=clipped_lon)
nc_load_clipped
```

```python
#convert to a dataframe
mus_df = nc_load_clipped.to_dataframe()
mus_df
```

```python
#read in the dataframe associated with the lat lon of the NADP sites
pathData = '/glade/u/home/demurray/External File Uploads'
os.chdir(pathData)
ntn = pd.read_csv('ntn.csv')
ntn['lon'] =  360 - (ntn['longitude'] * -1) #convert to same units as the model lon

#Clip ntn df to be the same range of lat and lon as the modelled data frame
ntn['latitude'] = ntn['latitude'].clip(lower=24, upper=50)
ntn['lon'] = ntn['lon'].clip(lower=235, upper=294)
ntn_loc = ntn[['siteId', 'latitude', 'lon']]
ntn_loc
```

```python
# Calculate distances between all points in mus_df and ntn_loc
distances = np.zeros((len(mus_df), len(ntn_loc)))

# Use tqdm to add a progress bar
for i, row in tqdm(mus_df.iterrows(), total=len(mus_df), desc='Calculating distances'):
    for j, row_ntn in ntn_loc.iterrows():
        distances[i, j] = haversine(row['lat'], row['lon'], row_ntn['latitude'], row_ntn['lon'])
```

```python
# Find the index of the closest point in mus_df for each point in ntn_loc
closest_indices = np.argmin(distances, axis=0)

# Keep only the latitude and longitude pairs in mus_df that correspond to the closest points in ntn_loc
mus_df_matched = mus_df.iloc[closest_indices]

# Assign the corresponding 'siteId' from ntn_loc to mus_df_matched
mus_df_matched['siteId'] = ntn_loc['siteId'].values
#mus_df_matched = mus_df_matched.rename(columns = {'lat': 'lat_mod', 'lon' : 'lon_mod'})
mus_df_matched.reset_index(inplace = True) #checks -- there should be the same length of cells in mus_df_matched as ntn_loc
mus_df_matched
```

```python
#from here, make a dictionary of site id that corresponds with model lat and lon and then 
#subset the xarray by this dictionary before perfroming any dataframe manipulation
result_dir =  Path("/glade/derecho/scratch/demurray/archive/MUSICA_trial_050824_01/atm/hist/")
file = "MUSICA_trial_050824_01.cam.h0.2017-01-01-00000.nc"
file_to_open = result_dir / file
nc_load_all = xr.open_dataset(file_to_open, engine='netcdf4')
nc_load_all
```

```python
# Step 1: Subset the xarray.Dataset using ncol
subset_ds = nc_load_all.sel(ncol=mus_df_matched['ncol'].values)

# Step 2: Assign siteId to the subset xarray.Dataset
# Convert DataFrame to xarray.Dataset for easier assignment
siteId_da = xr.DataArray(mus_df_matched['siteId'].values, dims=['ncol'], coords={'ncol': mus_df_matched['ncol'].values})
subset_ds = subset_ds.assign(siteId=siteId_da)
subset_ds
```

```python
#Merge df's to see how close the dfs are in terms of lat and lon
mus_ntn = pd.merge(mus_df_matched, ntn_loc, on = 'siteId', how = 'left') #looks good!
```

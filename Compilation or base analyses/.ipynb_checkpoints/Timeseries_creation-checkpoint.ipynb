```python
import pandas as pd
from pandas.tseries.offsets import DateOffset
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import griddata
import datetime as dt
from pathlib import Path
import os
from tqdm import tqdm
from fnmatch import fnmatch
```

```python
def process_daily_files(result_dir, files_to_open, variables_to_select, progress_bar=None):
    # Open the daily files as an xarray dataset
    ds = xr.open_mfdataset(str(result_dir / files_to_open), combine='nested')

    # Select specific variables
    ds = ds[variables_to_select]

    # Update progress bar
    if progress_bar:
        progress_bar.update(1)
    return ds
```

```python
# Extract SOA wet vars from newer model run (forgot to write some in the previous run)
# Define the directory, files to open, and variables to select
result_dir = Path('/glade/campaign/acom/acom-weather/demurray/FCnudged_f09.mam.Murray.Mar26_01.2002_2023.001.cam/atm_h0_SOAvars')
files_to_open = '*h0.*.nc'
pattern = '*SFWET'  # Adjust the pattern to match your variable names

# Get the list of files to process
files_list = list(result_dir.glob(files_to_open))

# Initialize tqdm progress bar
progress = tqdm(total=len(files_list))

# Processed data list
processed_data = []

# Iterate over each file
for file in files_list:
    # Open the file
    ds = xr.open_dataset(file)
    # Select variables that match the pattern
    selected_variables = [var for var in ds.variables if fnmatch(var, pattern)]
    # Close the dataset
    ds.close()
    # Process the selected variables
    processed_data.append(process_daily_files(result_dir, file.relative_to(result_dir), selected_variables))
    # Update the progress bar
    progress.update(1)

# Close the progress bar
progress.close()

# Concatenate the processed data along the 'time' dimension
nc_daily_other = xr.concat(processed_data, dim='time').sortby('time')
nc_daily_other
```

```python
# Extract SOA wet vars from newer model run (forgot to write some in the previous run)
# Define the directory, files to open, and variables to select
result_dir = Path('/glade/campaign/acom/acom-weather/demurray/FCnudged_f09.mam.Murray.Mar26_01.2002_2023.001.cam/atm_h0_SOAvars')
files_to_open = '*h0.*.nc'
pattern = '*DDF'  # Adjust the pattern to match your variable names

# Get the list of files to process
files_list = list(result_dir.glob(files_to_open))

# Initialize tqdm progress bar
progress = tqdm(total=len(files_list))

# Processed data list
processed_data = []

# Iterate over each file
for file in files_list:
    # Open the file
    ds = xr.open_dataset(file)
    # Select variables that match the pattern
    selected_variables = [var for var in ds.variables if fnmatch(var, pattern)]
    # Close the dataset
    ds.close()
    # Process the selected variables
    processed_data.append(process_daily_files(result_dir, file.relative_to(result_dir), selected_variables))
    # Update the progress bar
    progress.update(1)

# Close the progress bar
progress.close()

# Concatenate the processed data along the 'time' dimension
nc_daily_other = xr.concat(processed_data, dim='time').sortby('time')
nc_daily_other
```

```python
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np
import os
from tqdm import tqdm
import re
import numpy as np
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from pathlib import Path
from scipy.interpolate import griddata
```

```python
def plot_map_LCC(data, lat, lon, llim, ulim, units):
    #if you want to change the figure size of map projection you can do it here
    fig, ax = plt.subplots(figsize=(16, 9), subplot_kw=dict(projection=ccrs.LambertConformal()))
    #this is the actual plotting routine, using pcolormesh. Most variables are inputs from the function call
    im = ax.pcolormesh(lon,lat,data,cmap='ocean_r',vmin=llim, vmax=ulim, transform=ccrs.PlateCarree())
    #this is for visualization of administrative boundaries. Many shapes included in the 'Natural Earth' library can also be used
    ax.coastlines(resolution='50m')
    ax.add_feature(cfeature.BORDERS)
    ax.add_feature(cfeature.STATES, edgecolor='#D3D3D3')
    cb = fig.colorbar(im, ax=ax)
    cb.set_label(units)
    return fig, ax
```

```python
directory = '/glade/derecho/scratch/demurray/qfed_regrid_take2'

# Regular expression to capture the variable name from the filename
pattern = re.compile(r'qfed\.emis_(.*?)_hires_mol_\d{4}\.nc')

# Get a list of all files in the directory
all_files = os.listdir(directory)

# Use a set to store unique variables
unique_variables = set()

# Extract variable names from filenames
for file in all_files:
    match = pattern.match(file)
    if match:
        unique_variables.add(match.group(1))

# Print the unique variables
print("Unique variables:", unique_variables)
```

```python
# Process each unique variable with tqdm progress bar
for var in tqdm(unique_variables, desc="Processing variables"):
    file_2017 = os.path.join(directory, f'qfed.emis_{var}_hires_mol_2017.nc')
    file_2018 = os.path.join(directory, f'qfed.emis_{var}_hires_mol_2018.nc')
    
    if os.path.exists(file_2017) and os.path.exists(file_2018):
        # Read the NetCDF files
        ds_2017 = xr.open_dataset(file_2017)
        ds_2018 = xr.open_dataset(file_2018)
        
        # Concatenate along the time dimension
        ds_combined = xr.concat([ds_2017, ds_2018], dim='time')
        
        # Output file path
        output_file = os.path.join(directory, f'qfed.emis_{var}_hires_mol_2017_2018.nc')
        
        # Write out the combined dataset
        ds_combined.to_netcdf(output_file)
        
        print(f'Created {output_file}')
    else:
        print(f'One or both files for {var} do not exist.')
```
